```python
import re
```

```python
def normalize_arabic_text(text):
    # 1. Strip all diacritics (حركات)
    tashkeel = re.compile(r'[\u0617-\u061A\u064B-\u0652]')
    text = re.sub(tashkeel, '', text)
    
    # 2. Strip Tatweel (Kashida - the letter stretching character)
    text = re.sub(r'\u0640', '', text)
    
    # 3. Normalize ء variants to a bare 
    text = re.sub(r'[إأآا]', 'ا', text)
    
    # 4. Normalize Yaa and Alif Maqsura
    text = re.sub(r'[ىي]', 'ي', text)
    
    # 5. Normalize Taa Marbuta to Haa (Common in Gulf dialect transcription)
    # Alternatively, you can keep Taa Marbuta consistent depending on your labeling rules
    text = re.sub(r'ة', 'ه', text)
    
    # 6. Remove punctuation (optional but recommended for ASR baselines)
    punctuation = re.compile(r'[.,!?؛،؟"\'()[\]{}-]')
    text = re.sub(punctuation, ' ', text)
    
    # 7. Convert multiple spaces into a single space
    text = re.sub(r'\s+', ' ', text).strip()
    
    return text
```

```python
sample = "مَرْحَبَاً، شَخْبَارِكْ؟"
print(normalize_arabic_text(sample)) 
# Output should be clean: "مرحبا شخبارك"
```
